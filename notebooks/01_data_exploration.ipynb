```python
import pandas as pd
from pathlib import Path

# Convert the Windows file path into a Path object
data_path = Path(
    r"C:\Users\BHARATH KRISHNA P\Documents\European-Ecommerce-Analytics\data\raw\online_retail_II.xlsx"
)

print("Notebook directory:", Path.cwd())
print("Dataset location:", data_path)
print("File exists:", data_path.exists())
```

```python
excel_file = pd.ExcelFile(data_path)

print("Available sheets:")
print(excel_file.sheet_names)

for sheet in excel_file.sheet_names:
    df = pd.read_excel(
        data_path,
        sheet_name=sheet,
        nrows=5
    )

    print(f"\nSheet: {sheet}")
    print("Columns:", df.columns.tolist())
    display(df)
```

```python
dataframes = {}

for sheet in excel_file.sheet_names:
    df = pd.read_excel(data_path, sheet_name=sheet)
    dataframes[sheet] = df

    print(f"\n{'='*50}")
    print(f"Sheet: {sheet}")
    print(f"{'='*50}")

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())

    print("\nFirst five records:")
    display(df.head())
```

```python
for sheet, df in dataframes.items():
    print(f"\n--- {sheet} ---")

    if "Quantity" in df.columns:
        print(
            "Negative quantities:",
            (df["Quantity"] < 0).sum()
        )

    if "Price" in df.columns:
        print(
            "Zero or negative prices:",
            (df["Price"] <= 0).sum()
        )

    if "Country" in df.columns:
        print(
            "Countries:",
            df["Country"].nunique()
        )

        print("\nTop 10 countries:")
        print(df["Country"].value_counts().head(10))
```
